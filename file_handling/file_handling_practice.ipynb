### 3.6 Advanced: Reading Large Files Efficiently

```python
print("=" * 70)
print("PROCESSING LARGE FILES - MEMORY EFFICIENT")
print("=" * 70)

def process_large_csv(filename, chunk_size=1000):
    """Process a large CSV file in chunks to save memory"""
    total_rows = 0
    chunk_counter = 0
    
    with open(filename, 'r', encoding='utf-8') as file:
        csv_reader = csv.reader(file)
        header = next(csv_reader)
        
        for row in csv_reader:
            total_rows += 1
            # Process row here
            chunk_counter += 1
            
            if chunk_counter >= chunk_size:
                # Process chunk
                chunk_counter = 0
    
    return total_rows

# Test with sales data
row_count = process_large_csv('data/sales_data.csv')
print(f"📊 Total rows processed: {row_count}")
print("\n💡 Benefits of chunk processing:")
print("   ✓ Memory efficient (doesn't load entire file)")
print("   ✓ Can process files larger than RAM")
print("   ✓ Progress tracking possible")
print("   ✓ Early termination if needed")
```

### 4.2 Quoting Strategies

```python
print("=" * 70)
print("QUOTING STRATEGIES")
print("=" * 70)

data = [
    ['Product', 'Description', 'Price'],
    ['Laptop', 'High-performance, 16GB RAM', '1299.99'],
    ['Smartphone', '5G, 256GB storage', '899.50'],
    ['Headphones', 'Noise-cancelling, wireless', '199.95']
]

# Strategy 1: QUOTE_MINIMAL (default) - quotes only when necessary
with open('output/minimal_quotes.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f, quoting=csv.QUOTE_MINIMAL)
    writer.writerows(data)

# Strategy 2: QUOTE_ALL - quotes everything
with open('output/all_quotes.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f, quoting=csv.QUOTE_ALL)
    writer.writerows(data)

# Strategy 3: QUOTE_NONNUMERIC - quotes non-numeric fields
with open('output/non_numeric_quotes.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f, quoting=csv.QUOTE_NONNUMERIC)
    writer.writerows(data)

# Strategy 4: QUOTE_NONE - no quotes (may cause errors with special characters)
with open('output/no_quotes.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f, quoting=csv.QUOTE_NONE, escapechar='\\')
    writer.writerows(data)

# Display results
files = ['minimal_quotes', 'all_quotes', 'non_numeric_quotes', 'no_quotes']
for filename in files:
    print(f"\n📌 {filename}:")
    with open(f'output/{filename}.csv', 'r', encoding='utf-8') as f:
        print(f.read().strip())
```

### 5.3 Error Handling and Data Validation

```python
print("=" * 70)
print("ERROR HANDLING AND DATA VALIDATION")
print("=" * 70)

def safe_csv_reader(filename, expected_columns=None):
    """Safely read CSV with validation"""
    valid_rows = []
    error_rows = []
    
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            reader = csv.reader(file)
            header = next(reader)
            
            print(f"📋 Header: {header}")
            
            for i, row in enumerate(reader, 2):  # Start at row 2
                try:
                    # Validate row length
                    if expected_columns and len(row) != expected_columns:
                        raise ValueError(f"Expected {expected_columns} columns, got {len(row)}")
                    
                    # Validate no empty required fields
                    if any(field.strip() == '' for field in row[:2]):  # First 2 fields required
                        raise ValueError("Empty required field")
                    
                    valid_rows.append(row)
                    
                except ValueError as e:
                    error_rows.append((i, row, str(e)))
                    
    except FileNotFoundError:
        print(f"❌ Error: File '{filename}' not found")
        return [], []
    except Exception as e:
        print(f"❌ Unexpected error: {e}")
        return [], []
    
    return valid_rows, error_rows

# Test with incomplete data
print("🔍 Testing with incomplete data:\n")
valid, errors = safe_csv_reader('data/incomplete_data.csv', expected_columns=5)

print(f"\n✅ Valid rows: {len(valid)}")
for row in valid:
    print(f"  {row}")

if errors:
    print(f"\n❌ Error rows: {len(errors)}")
    for row_num, row, error in errors:
        print(f"  Row {row_num}: {row} -> {error}")

print("\n💡 Always validate data:")
print("   ✓ Check column count")
print("   ✓ Check for empty required fields")
print("   ✓ Convert and validate data types")
print("   ✓ Handle encoding issues")
```

### 5.4 CSV to Other Formats

```python
print("=" * 70)
print("CSV TO JSON CONVERTER")
print("=" * 70)

def csv_to_json(csv_filename, json_filename):
    """Convert CSV to JSON format"""
    data = []
    
    with open(csv_filename, 'r', encoding='utf-8') as csv_file:
        reader = csv.DictReader(csv_file)
        for row in reader:
            # Convert numeric fields
            for key, value in row.items():
                try:
                    if '.' in value:
                        row[key] = float(value)
                    else:
                        row[key] = int(value)
                except (ValueError, TypeError):
                    pass  # Keep as string
            data.append(row)
    
    with open(json_filename, 'w', encoding='utf-8') as json_file:
        json.dump(data, json_file, indent=2, ensure_ascii=False)
    
    return len(data)

# Convert employees.csv to JSON
count = csv_to_json('data/employees.csv', 'output/employees.json')
print(f"✅ Converted {count} records to JSON: output/employees.json")

# Display first few entries
with open('output/employees.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
    print(f"\n📄 JSON preview (first 2 records):")
    for i, record in enumerate(data[:2], 1):
        print(f"\n  Record {i}:")
        for key, value in record.items():
            print(f"    {key}: {value}")
```

### Exercise 1: Sales Data Analysis
**Task:** Analyze the sales data and generate a summary report

```python
def analyze_sales_data():
    """Analyze sales data and generate insights"""
    
    # Initialize counters
    product_sales = defaultdict(int)
    region_sales = defaultdict(int)
    total_revenue = 0
    total_quantity = 0
    
    with open('data/sales_data.csv', 'r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        
        for row in reader:
            product = row['Product']
            region = row['Region']
            quantity = int(row['Quantity'])
            price = float(row['Unit_Price'])
            
            revenue = quantity * price
            
            product_sales[product] += revenue
            region_sales[region] += revenue
            total_revenue += revenue
            total_quantity += quantity
    
    # Display results
    print("=" * 70)
    print("SALES ANALYSIS REPORT")
    print("=" * 70)
    
    print(f"\n💰 Total Revenue: ${total_revenue:,.2f}")
    print(f"📦 Total Quantity Sold: {total_quantity:,}")
    print(f"📊 Average Price per Unit: ${total_revenue/total_quantity:,.2f}")
    
    print("\n🏆 Top 5 Products by Revenue:")
    for product, revenue in sorted(product_sales.items(), key=lambda x: x[1], reverse=True)[:5]:
        print(f"  {product}: ${revenue:,.2f}")
    
    print("\n🌍 Revenue by Region:")
    for region, revenue in sorted(region_sales.items(), key=lambda x: x[1], reverse=True):
        percentage = (revenue / total_revenue) * 100
        print(f"  {region}: ${revenue:,.2f} ({percentage:.1f}%)")

analyze_sales_data()
```

```python
def process_student_grades():
    """Process student grades and calculate statistics"""
    
    subject_averages = defaultdict(list)
    student_results = []
    
    with open('data/students.csv', 'r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        subjects = ['Math', 'Science', 'English', 'History', 'Art']
        
        for row in reader:
            name = row['Name']
            grades = [int(row[subject]) for subject in subjects]
            avg_grade = sum(grades) / len(grades)
            
            student_results.append({
                'Name': name,
                'Average': round(avg_grade, 2),
                'Grade': 'A' if avg_grade >= 90 else 'B' if avg_grade >= 80 else 'C' if avg_grade >= 70 else 'D' if avg_grade >= 60 else 'F'
            })
            
            for subject, grade in zip(subjects, grades):
                subject_averages[subject].append(grade)
    
    # Display results
    print("=" * 70)
    print("STUDENT GRADE ANALYSIS")
    print("=" * 70)
    
    print("\n📊 Subject Averages:")
    for subject, grades in subject_averages.items():
        avg = sum(grades) / len(grades)
        print(f"  {subject}: {avg:.2f}")
    
    print("\n🏆 Top 5 Students:")
    for student in sorted(student_results, key=lambda x: x['Average'], reverse=True)[:5]:
        print(f"  {student['Name']}: {student['Average']} ({student['Grade']})")
    
    # Grade distribution
    grade_counts = Counter([s['Grade'] for s in student_results])
    print("\n📈 Grade Distribution:")
    for grade in ['A', 'B', 'C', 'D', 'F']:
        count = grade_counts.get(grade, 0)
        percentage = (count / len(student_results)) * 100
        print(f"  {grade}: {count} students ({percentage:.1f}%)")
    
    return student_results

students = process_student_grades()
```

### Exercise 3: CSV Data Cleaner
**Task:** Clean messy CSV data with inconsistent formatting

```python
def clean_messy_csv(input_file, output_file):
    """Clean messy CSV data"""
    
    cleaned_rows = []
    issues = []
    
    with open(input_file, 'r', encoding='utf-8') as file:
        reader = csv.reader(file)
        header = next(reader)
        cleaned_rows.append(header)
        
        for row_num, row in enumerate(reader, 2):
            cleaned_row = []
            
            for col_num, value in enumerate(row):
                # Clean value
                cleaned_value = value.strip()  # Remove whitespace
                
                # Handle missing values
                if cleaned_value == '' or cleaned_value == 'N/A' or cleaned_value == 'NULL':
                    cleaned_value = 'Unknown'
                    issues.append(f"Row {row_num}, Column {col_num+1}: Missing value replaced with 'Unknown'")
                
                # Clean price fields
                if '$' in cleaned_value:
                    cleaned_value = cleaned_value.replace('$', '')
                
                cleaned_row.append(cleaned_value)
            
            cleaned_rows.append(cleaned_row)
    
    # Write cleaned data
    with open(output_file, 'w', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        writer.writerows(cleaned_rows)
    
    return len(cleaned_rows) - 1, len(issues), issues[:5]  # Return summary

# Clean the messy data
rows_cleaned, issues_found, sample_issues = clean_messy_csv('data/messy_data.csv', 'output/cleaned_data.csv')

print("=" * 70)
print("DATA CLEANING RESULTS")
print("=" * 70)
print(f"✅ Rows processed: {rows_cleaned}")
print(f"🔧 Issues fixed: {issues_found}")
print(f"\n📝 Sample fixes:")
for issue in sample_issues:
    print(f"  • {issue}")

print(f"\n📄 Cleaned file: {output_file}")

# Preview cleaned data
print("\n📋 Preview of cleaned data:")
with open('output/cleaned_data.csv', 'r', encoding='utf-8') as f:
    for i in range(4):
        print(f"  {f.readline().strip()}")
```

### Exercise 4: CSV Merger
**Task:** Merge multiple CSV files with same structure

```python
def merge_csv_files(file_list, output_file):
    """Merge multiple CSV files with the same header"""
    
    merged_count = 0
    header_written = False
    
    with open(output_file, 'w', newline='', encoding='utf-8') as output:
        writer = csv.writer(output)
        
        for file_path in file_list:
            try:
                with open(file_path, 'r', encoding='utf-8') as input_file:
                    reader = csv.reader(input_file)
                    header = next(reader)
                    
                    # Write header only once
                    if not header_written:
                        writer.writerow(header)
                        header_written = True
                    
                    # Write data rows
                    for row in reader:
                        writer.writerow(row)
                        merged_count += 1
            except Exception as e:
                print(f"❌ Error processing {file_path}: {e}")
    
    return merged_count

# Create two CSV files to merge
with open('data/part1.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['ID', 'Name', 'Value'])
    writer.writerows([[1, 'A', 100], [2, 'B', 200]])

with open('data/part2.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['ID', 'Name', 'Value'])
    writer.writerows([[3, 'C', 300], [4, 'D', 400]])

# Merge files
files_to_merge = ['data/part1.csv', 'data/part2.csv']
total = merge_csv_files(files_to_merge, 'output/merged_files.csv')

print("=" * 70)
print("CSV MERGER RESULTS")
print("=" * 70)
print(f"✅ Merged {len(files_to_merge)} files")
print(f"📊 Total rows merged: {total}")
print(f"📄 Output: output/merged_files.csv")

# Preview merged file
print("\n📋 Merged content:")
with open('output/merged_files.csv', 'r', encoding='utf-8') as f:
    print(f.read())
```

### Exercise 5: CSV Filter and Query
**Task:** Implement a simple query system for CSV data

```python
def query_csv(filename, conditions=None, sort_by=None, limit=None):
    """
    Query CSV data with conditions
    
    Parameters:
        filename: CSV file path
        conditions: dict of column:value pairs to filter
        sort_by: column to sort by
        limit: max rows to return
    """
    results = []
    
    with open(filename, 'r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        
        for row in reader:
            # Apply conditions
            if conditions:
                match = True
                for col, value in conditions.items():
                    if col in row and row[col].strip() != str(value).strip():
                        match = False
                        break
                if not match:
                    continue
            
            results.append(row)
    
    # Sort results
    if sort_by and results:
        try:
            # Try numeric sort
            results.sort(key=lambda x: float(x.get(sort_by, 0)) if x.get(sort_by, '').replace('.', '').isdigit() else x.get(sort_by, ''))
        except:
            # Fallback to string sort
            results.sort(key=lambda x: x.get(sort_by, ''))
    
    # Apply limit
    if limit:
        results = results[:limit]
    
    return results

# Query examples
print("=" * 70)
print("CSV QUERY SYSTEM")
print("=" * 70)

# Query 1: Engineering employees
print("\n🔍 Query: Department = 'Engineering'")
results = query_csv('data/employees.csv', conditions={'Department': 'Engineering'})
for r in results:
    print(f"  {r['FirstName']} {r['LastName']} - {r['Department']} (${r['Salary']})")

# Query 2: High salary, sorted
print("\n🔍 Query: Top 3 highest paid employees")
results = query_csv('data/employees.csv', sort_by='Salary', limit=3)
for r in results:
    print(f"  {r['FirstName']} {r['LastName']}: ${r['Salary']}")

# Query 3: Active employees in Sales
print("\n🔍 Query: Active employees in Sales")
results = query_csv('data/employees.csv', conditions={'Department': 'Sales', 'Status': 'Active'})
for r in results:
    print(f"  {r['FirstName']} {r['LastName']} - {r['Department']}")
```

### Exercise 6: CSV to SQL
**Task:** Generate SQL INSERT statements from CSV data

```python
def csv_to_sql_insert(csv_filename, table_name, output_file):
    """Generate SQL INSERT statements from CSV"""
    
    with open(csv_filename, 'r', encoding='utf-8') as csv_file:
        reader = csv.DictReader(csv_file)
        columns = reader.fieldnames
        
        sql_statements = [f"-- SQL INSERT statements generated from {csv_filename}"]
        sql_statements.append(f"INSERT INTO {table_name} ({', '.join(columns)}) VALUES\n")
        
        values_list = []
        for row in reader:
            values = []
            for col in columns:
                value = row[col]
                # Escape quotes and handle NULLs
                if value == '' or value == 'NULL' or value == 'null':
                    values.append('NULL')
                else:
                    # Escape single quotes
                    value = value.replace("'", "''")
                    values.append(f"'{value}'")
            values_list.append(f"  ({', '.join(values)})")
        
        sql_statements.append(',\n'.join(values_list) + ';')
    
    # Write to file
    with open(output_file, 'w', encoding='utf-8') as f:
        f.write('\n'.join(sql_statements))
    
    return len(values_list)

# Generate SQL for employees
count = csv_to_sql_insert('data/employees.csv', 'employees', 'output/employees.sql')

print("=" * 70)
print("SQL GENERATOR RESULTS")
print("=" * 70)
print(f"✅ Generated {count} SQL INSERT statements")
print(f"📄 Output: output/employees.sql")

# Preview SQL
print("\n📋 SQL Preview:")
with open('output/employees.sql', 'r', encoding='utf-8') as f:
    for i, line in enumerate(f):
        if i < 5:  # Show first 5 lines
            print(f"  {line.strip()}")
        else:
            print("  ...")
            break
```
