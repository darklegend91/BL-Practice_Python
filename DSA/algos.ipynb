```python
def binary_search_iter(arr: list , target : int ) -> int: 
    """Iterative Approach Time Complexity : O(log n) and Space Complexity: O(1)"""
    left = 0
    right = len(arr) -1
    
    while left <= right:
        mid = (left + right) // 2
        
        if (arr[mid] == target): #If element is found
            return mid

        elif target < arr[mid]: #if target is small then start
            right = mid - 1
        
        else:
            left = mid + 1
    
    return -1


target_list = [12 , 45 ,67 ,89 ,104 , 380]
target_ele = 104

print(binary_search_iter(target_list,target_ele) +1 )
```
