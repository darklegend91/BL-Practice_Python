Import the numpy 

```python
import numpy as np
```

numoy arrays are more efficient for the numercical operations than the python lists

```python
python_list = [1,2,3,4,5]

numpy_array = np.array(python_list) # Make a numpy array using the python list

print("Python List : " , python_list)
print("Numpy array : " , numpy_array)

print("Type of Python List : " , type(python_list))
print("Type of Numpy array : " , type(numpy_array))

# adding 10 to each value of lost and ndarray
print("Python list (loop required)" , [x +10 for x in python_list])
print("Numpy Array (Vectorised)" , numpy_array + 10)
```

#### MAKE A NUMPY ARRAY

```python
#from an array
arr1 = np.array([1,2,3,4,5,6] , dtype = np.float32)

# from a range
arr2 = np.arange(0,10,2) 

#evenly spaced numbers
arr3 = np.linspace(0,1,5)

# zero array

arr4 = np.zeros(10)
arr5 = np.ones(10)
arr6 = np.full(5,7) #fill 5 length array with 7 value at each position

arr_iden = np.eye(5)

rand_arr = np.random.rand(5) #5 value

rand_arr_2 = np.random.randint(1,10 , size = 5) #[low , hight , size]
print(arr1)
print(arr2)
print(arr3)
print(arr4)
print(arr5)
print(arr6)
print(rand_arr)
print(rand_arr_2)
print(arr_iden)



print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)
print(arr4.dtype)
print(arr5.dtype)
print(arr6.dtype)
print(rand_arr.dtype)
print(rand_arr_2.dtype)
print(arr_iden.dtype)
```
